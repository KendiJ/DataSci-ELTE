```python
import pandas as pd
import numpy as np

from ucimlrepo import fetch_ucirepo
```

```python
# Load dataset
student_performance = fetch_ucirepo(id=320)
X = student_performance.data.features
y = student_performance.data.targets
df = pd.concat([X, y], axis=1)

print("Dataset loaded:", df.shape)
df.head(3)
```

## Exercise 1
- Compute the mean, median, and mode for the column absences.
- Is the mean or median more representative here? Why?

```python
col = "absences"

mean = df[col].mean()
median = df[col].median()
mode   = df[col].mode()[0]

print(f"Column : {col}")
print(f"Mean   : {mean:.2f}")
print(f"Median : {median:.2f}")
print(f"Mode   : {mode}")
print()
print("Interpretation:")
print(f"  Mean ≈ Median ({mean:.1f} ≈ {median:.1f}) → distribution is roughly symmetric")
```

## Exercise 2
- Compute the IQR for age.
- Identify any values that fall outside Q1 − 1.5×IQR or Q3 + 1.5×IQR — these are considered outliers by the Tukey rule.

## Hint:
- lower = q1 - 1.5 * iqr
- upper = q3 + 1.5 * iqr
- df[(df["age"] < lower) | (df["age"] > upper)]

```python
col = "age"
q1 = df[col].quantile(0.25)
q3 = df[col].quantile(0.75)

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

outliers = df[(df[col] < lower) | (df[col] > upper)]

print(f"Q1: {q1}, Q3: {q3}, IQR: {iqr}")
print(f"Lower: {lower}, Upper: {upper}")
print(f"Outliers found:")
print(outliers)

```
